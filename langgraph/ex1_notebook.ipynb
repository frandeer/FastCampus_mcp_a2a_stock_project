```python
from typing import TypedDict
from langgraph.graph import StateGraph
from langgraph.graph import START, END
```

## 1. 상태 정의
모든 노드에서 공유할 데이터 구조를 정의합니다.

```python
class State(TypedDict):
    message: str
    result: str
```

## 2. 노드 함수 정의
실제 작업을 수행하는 함수를 만듭니다.

```python
def process_node(state: State):
    return {"result": f"처리됨: {state['message']}"}
```

## 3. 그래프 생성
노드들을 연결해서 워크플로우를 만듭니다.

```python
graph = StateGraph(State)
graph.add_node("process", process_node)
graph.add_edge(START, "process")
graph.add_edge("process", END)
```

## 4. 그래프 시각화
만든 워크플로우의 구조를 확인해봅시다.

```python
app = graph.compile()

# 그래프 구조를 Mermaid 형식으로 출력
print("=== 그래프 다이어그램 ===")
print(app.get_graph().draw_mermaid())
```

## 5. 실행 및 결과 확인

```python
# 워크플로우 실행
result = app.invoke({"message": "Hello LangGraph"})
print("=== 실행 결과 ===")
print(result)
```

## 6. 다양한 입력으로 테스트

```python
test_messages = [
    "안녕하세요",
    "데이터 처리", 
    "AI 워크플로우"
]

for msg in test_messages:
    result = app.invoke({"message": msg})
    print(f"입력: {msg}")
    print(f"출력: {result['result']}")
    print("---")
```
